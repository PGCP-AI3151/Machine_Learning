```python
import pandas as pd
from sklearn.model_selection import KFold , StratifiedKFold, cross_val_score
from sklearn.ensemble import GradientBoostingClassifier

import warnings
warnings.filterwarnings('ignore')
```

```python
df = pd.read_csv(r"C:\Users\PGCP-AI.STUDENTSDC\Desktop\Machine_Learning\Datasets\pizza.csv")
df.sample(5)
```

```python
df = pd.read_csv(r"C:\Users\PGCP-AI.STUDENTSDC\Desktop\Machine_Learning\Cases\Vehicle Silhouettes\Vehicle.csv")
df.sample(5)
```

```python
from sklearn.preprocessing import LabelEncoder

X , y = df.drop('Class',axis=1) , df['Class']
encoder = LabelEncoder()
y_en = encoder.fit_transform(y)

kfold = StratifiedKFold(n_splits=5 , shuffle=True , random_state=26)
gbm = GradientBoostingClassifier(random_state=26)
```

### Accuracy Score

```python
results = cross_val_score(gbm,X,y_en,cv=kfold)  # default scoring='accuracy_score'
results
```

### Log Loss

```python
results = cross_val_score(gbm,X,y_en,cv=kfold,scoring='neg_log_loss')  
results
```

```python
import numpy as np
from tqdm import tqdm
from sklearn.linear_model import LogisticRegression

solvers = ['lbfgs', 'saga', 'sag', 'newton-cg', 'newton-cholesky']
scores = []

for s  in tqdm(solvers):
    lr = LogisticRegression(solver=s)
    kfold = StratifiedKFold(n_splits=5 , shuffle=True , random_state=26)
    results = cross_val_score(lr,X,y_en,cv=kfold,scoring='neg_log_loss')
    scores.append([s,np.mean(results)])
df_scores = pd.DataFrame(scores,columns=['Solver','Fold Log Loss'])
df_scores.sort_values('Fold Log Loss',ascending=False,ignore_index=True)
```

```python
import numpy as np
from tqdm import tqdm
from sklearn.linear_model import LogisticRegression

solvers = ['lbfgs', 'saga', 'sag', 'newton-cg', 'newton-cholesky']
Cs = np.linspace(0.01,3,5)
scores = []

for s  in tqdm(solvers):
    for c in Cs:
        lr = LogisticRegression(solver=s,C=c)
        kfold = StratifiedKFold(n_splits=5 , shuffle=True , random_state=26)
        results = cross_val_score(lr,X,y_en,cv=kfold,scoring='neg_log_loss')
        scores.append([s,c,np.mean(results)])
df_scores = pd.DataFrame(scores,columns=['Solver','C','Fold Log Loss'])
df_scores.sort_values('Fold Log Loss',ascending=False,ignore_index=True)
```
