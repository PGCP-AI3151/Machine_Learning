```python
import pandas as pd
import numpy as np
from sklearn.svm import SVC
from sklearn.model_selection import train_test_split
from sklearn.metrics import accuracy_score
from sklearn.preprocessing import OneHotEncoder, StandardScaler
from sklearn.pipeline import Pipeline
from sklearn.compose import make_column_selector , make_column_transformer

from sklearn.multiclass import OneVsOneClassifier
from tqdm import tqdm
```

```python
df = pd.read_csv(r"C:\Users\PGCP-AI.STUDENTSDC\Desktop\Machine_Learning\Cases\Vehicle Silhouettes\Vehicle.csv")
df.sample(5)
```

```python
X , y = df.drop('Class',axis=1), df['Class']
X_train,X_test,y_train,y_test = train_test_split(X,y,test_size=0.3,random_state=26,stratify=y)
```

### Kernel = linear

```python
cs = np.linspace(0.001,10,20)
scores = []
dfs = ['ovo','ovr']
scaler = StandardScaler()

for c in tqdm(cs) :
    for f in dfs:
        svm = SVC(kernel='linear',C=c,decision_function_shape=f)
        pipe = Pipeline([('SCL',scaler),('SVM',svm)])
        pipe.fit(X_train,y_train)
        y_pred = pipe.predict(X_test)
        scores.append([c,f,accuracy_score(y_test,y_pred)])
df_score = pd.DataFrame(scores,columns=['C','Decision','Accuracy'])
df_score.sort_values('Accuracy',ascending=False)
```

### Kernel = rbf(Radial Basis Funation)

```python
cs = np.linspace(0.001,10,20)
gs = np.linspace(0.1,10,20)
scores = []
dfs = ['ovo','ovr']
scaler = StandardScaler()

for c in tqdm(cs) :
    for g in gs:
        for f in dfs:
            svm = SVC(kernel='rbf',C=c,decision_function_shape=f,gamma=g)
            pipe = Pipeline([('SCL',scaler),('SVM',svm)])
            pipe.fit(X_train,y_train)
            y_pred = pipe.predict(X_test)
            scores.append([c,g,f,accuracy_score(y_test,y_pred)])
df_score = pd.DataFrame(scores,columns=['C','Gamma','Decision','Accuracy'])
df_score.sort_values('Accuracy',ascending=False)
```
