# boston.csv dataset

```python
df = pd.read_csv('boston.csv')
df.head()
```

```python
sns.heatmap(df.corr(),annot=True,cmap='Blues',fmt='.2f',annot_kws={'size':7},linewidths=0.5,linecolor='white')
plt.show()
```

```python
from sklearn.model_selection import train_test_split 
from sklearn.metrics import root_mean_squared_error
from sklearn.linear_model import LinearRegression
```

```python
X , y = df.drop('medv',axis=1),df['medv']
X_train, X_test, y_train, y_test = train_test_split(X, y, random_state=26, test_size=0.3)
model = LinearRegression()
model.fit(X_train, y_train)
```

```python
y_pred = model.predict(X_test)
```

```python
root_mean_squared_error(y_test, y_pred)
```
