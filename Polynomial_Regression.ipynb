```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
from sklearn.linear_model import LinearRegression
from sklearn.preprocessing import PolynomialFeatures
from sklearn.model_selection import train_test_split
from sklearn.metrics import root_mean_squared_error
```

```python
df = pd.read_csv(r"C:\Users\PATEL VIRAJ\OneDrive\Desktop\PGCP_AI\Machine_Learning\Datasets\pizza.csv")

X = df[['Promote']]
y = df['Sales']

poly = PolynomialFeatures(degree=3,include_bias=False).set_output(transform='pandas')
X_poly = poly.fit_transform(X)
X_poly
```

```python
X_train,X_test,y_train,y_test = train_test_split(X_poly,y,random_state=26,train_size=0.7)
model = LinearRegression()
model.fit(X_train,y_train)
y_pred = model.predict(X_test)
root_mean_squared_error(y_test,y_pred)
```

```python
print(model.coef_)
print(model.intercept_)
```

```python
import pandas as pd 
import matplotlib.pyplot as plt 
```

```python
df  = pd.read_csv(r'C:\Users\PATEL VIRAJ\OneDrive\Desktop\PGCP_AI\Machine_Learning\Datasets\Boston.csv')
df.head()
```

```python
X = df.drop('medv',axis=1)
y = df['medv']
```

```python
X_train,X_test,y_train,y_test = train_test_split(X,y,train_size=0.8,random_state=26)
```

```python
from sklearn.linear_model import LinearRegression

model = LinearRegression()
model.fit(X_train,y_train)
```

```python
from sklearn.metrics import root_mean_squared_error,r2_score
y_pred = model.predict(X_test)
print('R2 Score ->',r2_score(y_test,y_pred))
print('RMSE ->',root_mean_squared_error(y_test,y_pred))
```

```python
# Polynomial Regression
from sklearn.preprocessing import PolynomialFeatures

poly = PolynomialFeatures(degree = 3,include_bias = False).set_output(transform='pandas')
X_poly = poly.fit_transform(X)
```

```python
X_train,X_test,y_train,y_test = train_test_split(X_poly,y,train_size=0.8,random_state=26)
```

```python
model = LinearRegression()
model.fit(X_train,y_train)
```

```python
y_pred = model.predict(X_test)
print('R2 Score ->',r2_score(y_test,y_pred))
print('RMSE ->',root_mean_squared_error(y_test,y_pred))
```

```python
#  for degree 2,3,4,5
degrees = [2,3,4,5]
score = []
```

```python
for x in degrees:
    df  = pd.read_csv(r'C:\Users\PATEL VIRAJ\OneDrive\Desktop\PGCP_AI\Machine_Learning\Datasets\Boston.csv')
    X = df.drop('medv',axis=1)
    y = df['medv']
    poly = PolynomialFeatures(degree = x,include_bias = False).set_output(transform='pandas')
    X_poly = poly.fit_transform(X)
    X_train,X_test,y_train,y_test = train_test_split(X_poly,y,train_size=0.8,random_state=26)
    model = LinearRegression()
    model.fit(X_train,y_train)
    y_pred = model.predict(X_test)
    score.append([x,r2_score(y_test,y_pred),root_mean_squared_error(y_test,y_pred)])
```

```python
scores = pd.DataFrame(score,columns=['Degree','R2 Score','Score'])
scores
```
