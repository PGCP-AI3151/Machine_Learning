```python
df = pd.read_csv(r"C:\Users\PGCP-AI.STUDENTSDC\Desktop\Machine_Learning\Cases\human-resources-analytics\HR_comma_sep.csv")
df.sample(5)
```

```python
X ,y = df.drop('left',axis=1), df['left']
X_train,X_test,y_train,y_test = train_test_split(X,y,test_size=0.3,random_state=26,stratify=y)
```

## Kerne = Linear

```python
from sklearn.preprocessing import OneHotEncoder,StandardScaler
from sklearn.pipeline import Pipeline
from sklearn.compose import make_column_selector , make_column_transformer

encoder = OneHotEncoder(sparse_output=False, drop='first')
ct = make_column_transformer((encoder,make_column_selector(dtype_include='object')),
                             ('passthrough',make_column_selector(dtype_include='number')),
                             verbose_feature_names_out = False)

scaler = StandardScaler()
svm = SVC(kernel='linear')
```

```python
pipe = Pipeline([('OHE',ct),('SCL',scaler),('SVM',svm)])
pipe.fit(X_train,y_train)
```

```python
y_pred = pipe.predict(X_test)
accuracy_score(y_test,y_pred)
```

## Kernel = Poly

```python
svm = SVC(kernel='poly')
pipe = Pipeline([('OHE',ct),('SCL',scaler),('SVM',svm)])
pipe.fit(X_train,y_train)
```

```python
y_pred = pipe.predict(X_test)
accuracy_score(y_test,y_pred)
```

# Exercise

Kernel = Linear

```python
# Try for multiple values for c -> 0.001,10 20-Values
import numpy as np
from tqdm import tqdm
cs = np.linspace(0.001,10,20)
scores = []

for c in tqdm(cs) :
    svm = SVC(kernel='linear',C=c)
    pipe = Pipeline([('OHE',ct),('SCL',scaler),('SVM',svm)])
    pipe.fit(X_train,y_train)
    y_pred = pipe.predict(X_test)
    scores.append([c,accuracy_score(y_test,y_pred)])
df_score = pd.DataFrame(scores,columns=['C','Accuracy'])
df_score.sort_values('Accuracy',ascending=False)
```

Kernel = Poly

```python
cs = np.linspace(0.001,10,20)
scores = []

for c in tqdm(cs) :
    for d in [2,3,4]:
        svm = SVC(kernel='poly',C=c,degree=d)
        pipe = Pipeline([('OHE',ct),('SCL',scaler),('SVM',svm)])
        pipe.fit(X_train,y_train)
        y_pred = pipe.predict(X_test)
        scores.append([c,d,accuracy_score(y_test,y_pred)])
df_score = pd.DataFrame(scores,columns=['C','Degree','Accuracy'])
df_score.sort_values('Accuracy',ascending=False)
```

Kernel = rbf

```python
cs = np.linspace(0.001,10,20)
gs = np.linspace(0.1,10,20)
scores = []

for c in tqdm(cs) :
    for g in gs:
        svm = SVC(kernel='rbf',C=c,gamma=g)
        pipe = Pipeline([('OHE',ct),('SCL',scaler),('SVM',svm)])
        pipe.fit(X_train,y_train)
        y_pred = pipe.predict(X_test)
        scores.append([c,g,accuracy_score(y_test,y_pred)])
df_score = pd.DataFrame(scores,columns=['C','Gamma','Accuracy'])
df_score.sort_values('Accuracy',ascending=False)
```
