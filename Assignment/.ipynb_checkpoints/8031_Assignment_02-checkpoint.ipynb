```python
import warnings 
warnings.filterwarnings('ignore')
```

```python
import pandas as pd
import numpy as np
from sklearn.model_selection import train_test_split
from sklearn.neighbors import KNeighborsClassifier
from sklearn.linear_model import LogisticRegression
from sklearn.svm import SVC
from sklearn.discriminant_analysis import LinearDiscriminantAnalysis , QuadraticDiscriminantAnalysis
from sklearn.preprocessing import OneHotEncoder,StandardScaler
from sklearn.compose import make_column_transformer, make_column_selector
from sklearn.pipeline import Pipeline
from sklearn.metrics import accuracy_score
```

```python
df = pd.read_csv('drug200.csv')
df.sample(5)
```

```python
X , y = df.drop('Drug',axis=1) , df['Drug']
X_train,X_test,y_train,y_test = train_test_split(X,y,test_size=0.3,random_state=26,stratify=y)
```

```python
encoder = OneHotEncoder(sparse_output = False, drop='first')
ct = make_column_transformer((encoder,make_column_selector(dtype_include='object')),
                             ('passthrough',make_column_selector(dtype_include='number')),
                             verbose_feature_names_out = False)
scaler = StandardScaler()
```

```python
models = {
    'Model':[],
    'Accuracy':[]
}
```

# KNeighborsClassifier

```python
ks = [2,3,5,7,8]
scores = []
for k in ks:
    knn = KNeighborsClassifier(n_neighbors=k)
    pipe = Pipeline([('OHE',ct),('SCL',scaler),('KNN',knn)])
    pipe.fit(X_train,y_train)
    y_pred = pipe.predict(X_test)
    scores.append([k,accuracy_score(y_test,y_pred)])
knn_score = pd.DataFrame(scores,columns=['K','Accuracy']).sort_values('Accuracy',ascending=False,
                                                                      ignore_index=True)
models['Model'].append('KNeighborsClassifier')
models['Accuracy'].append(knn_score['Accuracy'].iloc[0])
knn_score
```

# LogisticRegression

```python
solvers = ['lbfgs','liblinear','newton-cg','newton-cholesky','sag','saga']
scores = []
for s in solvers:
    lr = LogisticRegression(solver=s)
    pipe = Pipeline([('OHE',ct),('SCL',scaler),('LR',lr)])
    pipe.fit(X_train,y_train)
    y_pred = pipe.predict(X_test)
    scores.append([s,accuracy_score(y_test,y_pred)])
lr_score = pd.DataFrame(scores,columns=['Solver','Accuracy']).sort_values('Accuracy',ascending=False,
                                                                      ignore_index=True)
models['Model'].append('LogisticRegression')
models['Accuracy'].append(lr_score['Accuracy'].iloc[0])
lr_score
```

# SVC

```python
cs = np.linspace(0.001,10,20)
gs = np.linspace(0.1,10,20)
kernels = ['rbf','poly','linear','sigmoid']    
scores = []
for c in Cs:
    for g in gs:
        for k in kernels:
            svc = SVC(kernel=k , C=c, gamma=g)
            pipe = Pipeline([('OHE',ct),('SCL',scaler),('SVC',svc)])
            pipe.fit(X_train,y_train)
            y_pred = pipe.predict(X_test)
            scores.append([k,c,g,accuracy_score(y_test,y_pred)])
svc_score = pd.DataFrame(scores,columns=['Kernel','C','Gamma','Accuracy']).sort_values('Accuracy',ascending=False,
                                                                      ignore_index=True)
models['Model'].append('SVC')
models['Accuracy'].append(svc_score['Accuracy'].iloc[0])
svc_score
```

# LinearDiscriminantAnalysis

```python
solvers = ['lsqr', 'eigen']
shrinkage = [None, 'auto', 0.01, 0.1, 0.5, 0.9]
scores = []
for s in solvers:
    for sn in shrinkage:
        lda = LinearDiscriminantAnalysis(solver=s,shrinkage=sn)
        pipe = Pipeline([('OHE',ct),('SCL',scaler),('LDA',lda)])
        pipe.fit(X_train,y_train)
        y_pred = pipe.predict(X_test)
        scores.append([s,sn,accuracy_score(y_test,y_pred)])
lda_score = pd.DataFrame(scores,columns=['Solvers','Shrinkage','Accuracy']).sort_values('Accuracy',ascending=False, 
                                                                                        ignore_index=True)
models['Model'].append('LinearDiscriminantAnalysis')
models['Accuracy'].append(lda_score['Accuracy'].iloc[0])
lda_score
```

# QuadraticDiscriminantAnalysis

```python
reg_params = [0.0, 0.01, 0.1, 0.2, 0.5, 0.8, 1.0]
scores = []
for r in reg_params:
    qda = QuadraticDiscriminantAnalysis(reg_param=r)
    pipe = Pipeline([('OHE',ct),('SCL',scaler),('QDA',qda)])
    pipe.fit(X_train,y_train)
    y_pred = pipe.predict(X_test)       
    scores.append([r,accuracy_score(y_test,y_pred)])
qda_score = pd.DataFrame(scores,columns=['Reg_Param','Accuracy']).sort_values('Accuracy',ascending=False,
                                                                            ignore_index=True)
models['Model'].append('QuadraticDiscriminantAnalysis')
models['Accuracy'].append(qda_score['Accuracy'].iloc[0])
qda_score
```

```python
df_model = pd.DataFrame(models).sort_values('Accuracy',ascending=False,ignore_index=True)
df_model
```

```python
import matplotlib.pyplot as plt
bar = plt.bar(df_model['Model'],df_model['Accuracy'],color='purple',edgecolor='black')
ax = plt.gca()
ax.bar_label(bar)
plt.xticks(df_model['Model'],['SVC','Logistic','LDA','QDA','KNN'])
plt.title('Model Comparision',fontweight='bold')
plt.xlabel('Model',fontweight='bold')
plt.ylabel('Accuracy',fontweight='bold')
plt.show()
```
