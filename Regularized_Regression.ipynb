### Ridge Regression

```python
import pandas as pd
import numpy as np
from sklearn.preprocessing import OneHotEncoder
from sklearn.linear_model import Ridge
from sklearn.model_selection import train_test_split
from sklearn.metrics import root_mean_squared_error,r2_score
```

```python
df = pd.read_csv(r"C:\Users\PGCP-AI.STUDENTSDC\Desktop\Machine_Learning\Datasets\housing.csv")
df.head()
```

```python
X = df.drop('price',axis=1)
y = df['price']
```

```python
encoder = OneHotEncoder(sparse_output = False,drop='first').set_output(transform='pandas')
numeric = X.select_dtypes(include='number')
category = X.select_dtypes(include=['object','str'])
X_encoded = encoder.fit_transform(category)
X_encoded = pd.concat([numeric,X_encoded],axis=1)
X_encoded.head()
```

```python
X_train,X_test,y_train,y_test = train_test_split(X_encoded,y,train_size=0.7,random_state=26)
```

```python
model = Ridge() # default alpha = 1
model.fit(X_train,y_train)
model.coef_
```

```python
y_pred = model.predict(X_test)
print('R2 Score ->',r2_score(y_test,y_pred))
print('RMSE ->',root_mean_squared_error(y_test,y_pred))
```

```python
# Try For alpha = [0.1,0.25,0.7,1,1.5,2,5,10,50]
alpha = [0.1,0.25,0.7,1,1.5,2,5,10,50]
score =[]
```

```python
for x in alpha:
    model = Ridge(alpha=x) 
    model.fit(X_train,y_train)
    y_pred = model.predict(X_test)
    print('R2 Score ->',r2_score(y_test,y_pred))
    print('RMSE ->',root_mean_squared_error(y_test,y_pred))
    score.append([x,r2_score(y_test,y_pred),root_mean_squared_error(y_test,y_pred)])
```

```python
scores = pd.DataFrame(score , columns=['Alpha','R2 Score','RMSE'])
scores
```

```python
scores.sort_values('RMSE')
```

# Lasso Regression

```python
import pandas as pd
import numpy as np
from sklearn.linear_model import Lasso
from sklearn.model_selection import train_test_split
from sklearn.metrics import r2_score,root_mean_squared_error
```

```python
# Create dataset
from sklearn.datasets import make_regression
X, y = make_regression(
    n_samples=300,
    n_features=8,
    n_informative=3,
    noise=15,
    random_state=42
)

feature_names = [f"Feature_{i}" for i in range(1, 9)]
```

```python
X_train,X_test,y_train,y_test = train_test_split(X,y,random_state=42,train_size=0.8)
```

```python
model =  Lasso()
```

```python
# Start Class Work From Here 
```

```python
alpha = [0.1,0.25,0.7,1,1.5,2,5,10,50]
score =[]
for x in alpha:
    model = Lasso(alpha=x) 
    model.fit(X_train,y_train)
    y_pred = model.predict(X_test)
    score.append([x,r2_score(y_test,y_pred),root_mean_squared_error(y_test,y_pred)])
df_score = pd.DataFrame(score, columns=['Alpha','R2 Score','RMSE'])
df_score.sort_values('RMSE')
```

# Elastic Net Regression

```python
from sklearn.linear_model import ElasticNet
```

```python
alphas = [0.1,0.25,0.7,1,1.5,2,5,10,50]
ratios = [0.1,0.25,0.5,0.75,1]
scores = []
for a in alphas:
    for r in ratios:
        el = ElasticNet(alpha=a,l1_ratio=r)
        el.fit(X_train,y_train)
        y_pred = el.predict(X_test)
        scores.append([a,r,root_mean_squared_error(y_test,y_pred)])
df_score = pd.DataFrame(scores, columns=['Alpha','l1_ratio','RMSE'])
df_score.sort_values('RMSE')
```
