```python
from mlxtend.frequent_patterns import apriori
from mlxtend.frequent_patterns import association_rules

import pandas as pd
```

```python
df = pd.read_csv(r"C:\Users\PGCP-AI.STUDENTSDC\Desktop\Machine_Learning\Datasets\Faceplate.csv",index_col=0)
df.head()
```

```python
df = df.astype(bool)
df.info()
```

```python
itemset = apriori(df,min_support=0.2,use_colnames=True)
itemset
```

```python
rules = association_rules(itemset,metric='confidence',min_threshold=0.5)
rules = rules[['antecedents','consequents','support','confidence','lift']]
rules.sort_values(by=['confidence','lift','support'],ascending=False)
```

```python
relv_rules = rules[rules['lift']>1]
relv_rules.sort_values(by=['confidence','lift','support'],ascending=False)
```

# TransactionEncoder

```python
groceries = []
with open(r"C:\Users\PGCP-AI.STUDENTSDC\Desktop\Machine_Learning\Datasets\Groceries.csv",'r') as f:groceries = f.read()
groceries = groceries.split('\n')
groceries_list = []
for i in groceries:
    groceries_list.append(i.split(','))
```

```python
groceries_list   # list of list because TransactionEncoder required input as list of list
```

```python
from mlxtend.preprocessing import TransactionEncoder
te = TransactionEncoder()
te_ary = te.fit(groceries_list).transform(groceries_list)
df = pd.DataFrame(te_ary,columns=te.columns_)
df
```

```python
def gen_rules(min_support,min_confidence):
    itemsets = apriori(df,min_support=min_support,use_colnames=True)
    rules = association_rules(itemsets,metric='confidence',min_threshold=min_confidence)
    rules = rules[['antecedents','consequents','support','confidence','lift']]
    relv_rules = rules[rules['lift']>1]
    return relv_rules.sort_values(by=['confidence','lift','support'],ascending=False)
```

```python
interact(gen_rules,
         min_support = widgets.FloatSlider(value=0.01,min=0.01,max=0.1,step=0.01,description='Min Support:'),
         min_confidence = widgets.FloatSlider(value=0.01,min=0,max=1,step=0.01,description='Min Confidence:'))
```
