```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LinearRegression
```

```python
df = pd.read_csv(r"C:\Users\PGCP-AI.STUDENTSDC\Desktop\Machine_Learning\Datasets\pizza.csv")
df.head()
```

```python
df['Promote'].corr(df['Sales'])
```

```python
plt.scatter(df['Promote'],df['Sales'])
plt.title('Promotional Expense Vs Product Sales',fontweight='bold')
plt.xlabel('Promotional Expenses',fontweight='bold')
plt.ylabel('Sales',fontweight='bold')
plt.show()
```

```python
X = df[['Promote']]
y = df['Sales']
```

# Without Train Test Split

```python
model = LinearRegression()
model.fit(X,y)
```

```python
print('Coef :-',model.coef_)  # slope 
print('Intercept :-',model.intercept_)  # bias or weight or else -> ?
```

```python
promo = np.array([34,89,100,22])
sales_pred = 5.4858653632529695 + 23.50640302*promo
sales_pred
```

```python
tst = pd.DataFrame({'Promote': promo})
model.predict(tst)
```

# With Train Test Split

```python
X_train,X_test,y_train,y_test = train_test_split(X,y,train_size=0.8,random_state=42)
```

```python
model = LinearRegression()
model.fit(X_train,y_train)
```

```python
y_pred = model.predict(X_test)
y_pred = pd.DataFrame(y_pred.round(2))
y_pred
```

```python
y_test.iloc[:5]
```

```python
y_pred.iloc[:5]
```

```python
plt.scatter(X_train,y_train,color='g')
plt.plot(X_train,model.predict(X_train),color='r')
plt.title('Best Fit Line (Training Data)')
plt.show()
```

```python
plt.scatter(X_test,y_test,color='g')
plt.plot(X_test,model.predict(X_test),color='r')
plt.title('Best Fit Line (Training Data)')
plt.show()
```

```python
from sklearn.metrics import r2_score,mean_absolute_error
print('R2 :-',r2_score(y_test,y_pred))
print('Mean Absolute Error :-',mean_absolute_error(y_test,y_pred))
```
