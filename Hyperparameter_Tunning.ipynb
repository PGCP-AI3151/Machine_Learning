# GridSeachCV()

```python
import pandas as pd
from sklearn.model_selection import KFold , StratifiedKFold, GridSearchCV
from sklearn.ensemble import GradientBoostingClassifier


import warnings
warnings.filterwarnings('ignore')
```

```python
df = pd.read_csv(r"C:\Users\PGCP-AI.STUDENTSDC\Desktop\Machine_Learning\Cases\Vehicle Silhouettes\Vehicle.csv")
df.sample(5)
```

```python
from sklearn.preprocessing import LabelEncoder
from sklearn.linear_model import LogisticRegression
import numpy as np

X , y = df.drop('Class',axis=1) , df['Class']
encoder = LabelEncoder()
y_en = encoder.fit_transform(y)
kfold = StratifiedKFold(n_splits=5 , shuffle=True , random_state=26)
lr = LogisticRegression()
```

```python
params = {
    'solver':['lbfgs', 'saga', 'sag', 'newton-cg', 'newton-cholesky'],
    'C' : np.linspace(0.01,3,5)
}
gcv = GridSearchCV(lr,param_grid=params,cv=kfold, scoring='neg_log_loss')
gcv.fit(X,y_en)
```

```python
print('Best Params ->',gcv.best_params_)
print('Best Score ->',gcv.best_score_)
```

# Exercise With Pipeline

```python
df = pd.read_csv(r"C:\Users\PGCP-AI.STUDENTSDC\Desktop\Machine_Learning\Cases\Medical_Cost_Personal\insurance.csv")
df.sample(5)
```

```python
X , y = df.drop('charges',axis=1) , df['charges']
```

```python
from sklearn.compose import make_column_transformer, make_column_selector
from sklearn.preprocessing import OneHotEncoder
from sklearn.pipeline import Pipeline

encoder = OneHotEncoder(sparse_output=False, drop='first')
ct = make_column_transformer((encoder,make_column_selector(dtype_include='object')),
                             ('passthrough',make_column_selector(dtype_include='number')),
                             verbose_feature_names_out=False)

gbm = XGBRegressor(random_state=26)
pipe = Pipeline([('OHE',ct),('GBM',gbm)])
# pipe.get_params()
```

```python
params = {
    'GBM__learning_rate' : np.linspace(0.01,0.9,10),
    'GBM__max_depth':[None,2,3,4],
    'GBM__n_estimators':[25,50,100]
}

kfold = KFold(n_splits=5, shuffle=True, random_state=26)
gcv = GridSearchCV(pipe, param_grid=params,cv=kfold,scoring='r2')
gcv.fit(X,y)
```

```python
print('Best Params ->',gcv.best_params_)
print('Best Score ->',gcv.best_score_)
```

# Verbose

# n_jobs 
default=None\
n_jobs = -1 -> to speeding the run\
note -> don't use with verbose

```python
gcv = GridSearchCV(pipe, param_grid=params,cv=kfold,scoring='r2',n_jobs=-1)
gcv.fit(X,y)
```

```python
df = pd.read_csv(r"C:\Users\PGCP-AI.STUDENTSDC\Desktop\Machine_Learning\Cases\Chemical_Process_Data\ChemicalProcess.csv")
df.sample(5)
```

```python
df.isnull().sum().sum()
```

```python
X , y = df.drop('Yield',axis=1) , df['Yield']
```

```python
from sklearn.neighbors import KNeighborsRegressor
from sklearn.impute import SimpleImputer
from sklearn.preprocessing import StandardScaler , MinMaxScaler

impute = SimpleImputer()
scl_scaler = StandardScaler()
mm_scaler = MinMaxScaler()
knn = KNeighborsRegressor()

kfold = KFold(n_splits=5, shuffle=True, random_state=26)

pipe = Pipeline([('IMP',impute),('SCL',None),('KNN',knn)])
# pipe.get_params()
```

```python
params = {'IMP__strategy':['mean','median'],'SCL':[scl_scaler,mm_scaler,None],
          'KNN__n_neighbors':np.arange(1,11)}
gcv = GridSearchCV(pipe, param_grid=params, cv=kfold, scoring='r2', n_jobs=-1)
gcv.fit(X,y)
```

```python
print('Best Params ->',gcv.best_params_)
print('Best Score ->',gcv.best_score_)
```

```python
from ucimlrepo import fetch_ucirepo 
  
# fetch dataset 
cirrhosis_patient_survival_prediction = fetch_ucirepo(id=878) 
  
# data (as pandas dataframes) 
X = cirrhosis_patient_survival_prediction.data.features 
y = cirrhosis_patient_survival_prediction.data.targets 
  
# # metadata 
# print(cirrhosis_patient_survival_prediction.metadata) 
  
# # variable information 
# print(cirrhosis_patient_survival_prediction.variables) 
```

```python
X.sample(5)
```

```python
y.sample(5)
```

```python
X.columns
```

```python
X = X.replace('NaNN', np.nan)
X['Cholesterol'] = X['Cholesterol'].astype(float)
X['Copper'] = X['Copper'].astype(float)
X['Tryglicerides'] = X['Tryglicerides'].astype(float)
X['Platelets'] = X['Platelets'].astype(float)
```

```python
X.info()
```

```python
encoder = OneHotEncoder(sparse_output=False, drop='first')

ct = make_column_transformer((encoder,make_column_selector(dtype_include='object')),
                             ('passthrough',make_column_selector(dtype_include='number')),
                             verbose_feature_names_out=False).set_output(transform='pandas')
```

```python
num_imp = SimpleImputer()
cat_imp = SimpleImputer(strategy='constant',fill_value='Unkown')

ct_impute= make_column_transformer((num_imp,make_column_selector(dtype_include='number')),
                             (cat_imp,make_column_selector(dtype_include='object')),
                             verbose_feature_names_out=False).set_output(transform='pandas')
```

```python
from sklearn.neighbors import KNeighborsClassifier
scl_scaler = StandardScaler()
mm_scaler = MinMaxScaler()
knn = KNeighborsClassifier()

kfold = KFold(n_splits=5, shuffle=True, random_state=26)

pipe = Pipeline([('IMP',ct_impute),('OHE',ct),('SCL',None),('KNN',knn)])
# pipe.get_params()
```

```python
params = {
    'IMP__simpleimputer-1__strategy':['mean','median'],
    'IMP__simpleimputer-2__strategy':['constant','most_frequent'],
    'SCL':[scl_scaler,mm_scaler,None],
    'KNN__n_neighbors':np.arange(1,11)
}
gcv = GridSearchCV(pipe, param_grid=params, cv=kfold, scoring='neg_log_loss', n_jobs=-1)
gcv.fit(X,y)
```

```python
print('Best Params ->',gcv.best_params_)
print('Best Score ->',gcv.best_score_)
```

# With RandomForestClassifier()

```python
from sklearn.ensemble import RandomForestClassifier

forest = RandomForestClassifier(random_state=26)
pipe = Pipeline([('IMP',ct_impute),('OHE',ct),('RF',forest)])
pipe.get_params()
```

```python
params = {
    'IMP__simpleimputer-1__strategy':['mean','median'],
    'IMP__simpleimputer-2__strategy':['constant','most_frequent'],
    'RF__max_depth':[None,4,5,6,7],
    'RF__max_samples':[150,200,300,500],
    'RF__n_estimators':[25,50,100,150,200]
    
}
kfold = StratifiedKFold(n_splits=5, shuffle=True, random_state=26)
gcv = GridSearchCV(pipe, param_grid=params, cv=kfold, scoring='neg_log_loss', n_jobs=-1)
gcv.fit(X,y)
```

```python
print('Best Params ->',gcv.best_params_)
print('Best Score ->',gcv.best_score_)
```

# RandomizedSearchCV()

```python
from sklearn.model_selection import RandomizedSearchCV
from sklearn.preprocessing import OrdinalEncoder
from lightgbm import LGBMRegressor
```

```python
from ISLP import load_data
df = load_data('Carseats')
df.columns
```

```python
X , y = df.drop('Sales',axis=1) , df['Sales']
X_train,X_test,y_train,y_test = train_test_split(X,y,random_state=26,test_size=0.2,stratify=df['ShelveLoc'])
```

```python
ord_encoder = OrdinalEncoder(categories=[['Bad','Medium','Good']]).set_output(transform='pandas')
ohe_encoder = OneHotEncoder(sparse_output=False, drop='first')
ct= make_column_transformer((ord_encoder,['ShelveLoc']),
                            (ohe_encoder,['Urban','US']),
                             ('passthrough',make_column_selector(dtype_include='number')),
                             verbose_feature_names_out=False)
```

```python
lgbm = LGBMRegressor(random_state=26,verbose=-1)
pipe = Pipeline([('OHE',ct),('LGBM',lgbm)])

params = {
    'LGBM__max_depth':[None,3,4,5,7],
    'LGBM__learning_rate':np.linspace(0.01,0.9,10),
    'LGBM__n_estimators':[25,50,100]
}

rcv = RandomizedSearchCV(pipe,param_distributions=params,cv=kfold,scoring='r2',n_jobs=-1)
rcv.fit(X_train,y_train)
```
