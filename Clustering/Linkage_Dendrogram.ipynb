```python
from scipy.cluster.hierarchy import linkage,dendrogram

import pandas as pd
from sklearn.preprocessing import StandardScaler
import matplotlib.pyplot as plt
```

```python
df = pd.read_csv(r"C:\Users\PGCP-AI.STUDENTSDC\Desktop\Machine_Learning\Datasets\simp_clust.csv",index_col=0)
df.sample(5)
```

```python
plt.scatter(df['X1'],df['X2'])
for i in range(0,df.shape[0]):
    plt.text(x=df['X1'].values[i],
             y=df['X2'].values[i],
             s=list(df.index)[i])
plt.show()
```

```python
scaler = StandardScaler().set_output(transform='pandas')
df_scal = scaler.fit_transform(df)
link = 'average'
mergings = linkage(df_scal,method=link)
dendrogram(mergings, labels=list(df_scal.index))
plt.title(link+' linkage')
plt.show()
```

```python
link = 'complete'
mergings = linkage(df_scal,method=link)
dendrogram(mergings, labels=list(df_scal.index))
plt.title(link+' linkage')
plt.show()
```

```python
link = 'single'
mergings = linkage(df_scal,method=link)
dendrogram(mergings, labels=list(df_scal.index))
plt.title(link+' linkage')
plt.show()
```

```python
link = 'centroid'
mergings = linkage(df_scal,method=link)
dendrogram(mergings, labels=list(df_scal.index))
plt.title(link+' linkage')
plt.show()
```

```python
link = 'ward'
mergings = linkage(df_scal,method=link)
dendrogram(mergings, labels=list(df_scal.index))
plt.title(link+' linkage')
plt.show()
```

# ON Milk.csv

```python
df = pd.read_csv(r"C:\Users\PGCP-AI.STUDENTSDC\Desktop\Machine_Learning\Datasets\milk.csv",index_col=0)
df.sample(5)
```

```python
scaler = StandardScaler().set_output(transform='pandas')
df_scale = scaler.fit_transform(df)
merging = linkage(df_scale,method='average')
dendrogram(merging,labels=df_scale.index)
plt.title('Average Linkage')
plt.xticks(rotation=90)
plt.show()
```

```python
merging = linkage(df_scale,method='centroid')
dendrogram(merging,labels=df_scale.index)
plt.title('Centroid Linkage')
plt.xticks(rotation=90)
plt.show()
```

```python
merging = linkage(df_scale,method='single')
dendrogram(merging,labels=df_scale.index)
plt.title('Single Linkage')
plt.xticks(rotation=90)
plt.show()
```

```python
merging = linkage(df_scale,method='complete')
dendrogram(merging,labels=df_scale.index)
plt.title('Complete Linkage')
plt.xticks(rotation=90)
plt.show()
```

```python
merging = linkage(df_scale,method='ward')
dendrogram(merging,labels=df_scale.index)
plt.title('Ward Linkage')
plt.xticks(rotation=90)
plt.show()
```
