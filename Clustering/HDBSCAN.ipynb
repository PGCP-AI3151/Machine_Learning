```python
from sklearn.cluster import HDBSCAN

import pandas as pd
from sklearn.preprocessing import StandardScaler
import matplotlib.pyplot as plt
from tqdm import tqdm
from sklearn.metrics import silhouette_score

import warnings 
warnings.filterwarnings('ignore')
```

```python
df = pd.read_csv(r"C:\Users\PGCP-AI.STUDENTSDC\Desktop\Machine_Learning\Datasets\milk.csv",index_col=0)
df.sample(5)
```

```python
scaler = StandardScaler().set_output(transform='pandas')
df_scale = scaler.fit_transform(df)
```

```python
dbc = HDBSCAN(min_cluster_size=3)
dbc.fit(df_scale)
```

```python
inliners_obs = df_scale[dbc.labels_ != -1]
inliners_lbls = dbc.labels_[dbc.labels_ != -1]
silhouette_score(inliners_obs,inliners_lbls)
```
