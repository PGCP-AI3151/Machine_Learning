```python
import pandas as pd
import numpy as np
from tqdm import tqdm
from sentence_transformers import SentenceTransformer
```

```python
df = pd.read_csv('/content/titles.csv')
df.sample(5)
```

```python
df.shape
```

```python
ind_movie = df[(df['production_countries'].str.contains('IN')) & (df['type'] == 'MOVIE')]
ind_movie.head()
```

```python
ind_movie.shape
```

```python
us_movie = df[(df['production_countries'].str.contains('US')) & (df['type'] == 'MOVIE')]
us_movie.head()
```

```python
us_movie.shape
```

```python
model = SentenceTransformer('BAAI/bge-small-en-v1.5')
```

```python
all_desc = list(ind_movie['description'])
out_embs = []
for r in tqdm(all_desc):
  embd = model.encode(r)
  out_embs.append(embd)
out_embs = np.array(out_embs)
```

```python
out_embs.shape
```

```python
ind_movie['MovieID'] = ind_movie['id']+'-'+ind_movie['title']
df_embd = pd.DataFrame(out_embs, index=list(ind_movie['MovieID']))
df_embd
```

```python
from sklearn.cluster import HDBSCAN
from sklearn.metrics import silhouette_score
```

```python
hdbscan = HDBSCAN(min_cluster_size=2)
hdbscan.fit(out_embs)
```

```python
df_embd['hdbscan_cluster'] = hdbscan.labels_
df_embd['hdbscan_cluster'].value_counts()
```

```python
ind_movie[(df_embd['hdbscan_cluster'] == 48).values][['title','description']]
```

```python
df = pd.read_csv('/content/historic_articles.csv')
df.head()
```

```python
df.isna().sum()
df.dropna(inplace=True)
```

```python
all_content = list(df['content'])
embd = []
for c in tqdm(all_content):
  em = model.encode(c)
  embd.append(em)
embd = np.array(embd)
```

```python
df_embd = pd.DataFrame(embd)
df_embd
```

```python
hdbscan = HDBSCAN(min_cluster_size=2)
hdbscan.fit(embd)
```

```python
df_embd['Cluster'] = hdbscan.labels_
df_embd['Cluster'].value_counts()
```

```python
df[(df_embd['Cluster'] == 98).values][['source','category','content']]
```
