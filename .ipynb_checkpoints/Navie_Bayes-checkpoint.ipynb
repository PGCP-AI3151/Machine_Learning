# BernoulliNB

```python
import pandas as pd
import numpy as np
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import OneHotEncoder
from sklearn.pipeline import Pipeline
from sklearn.compose import make_column_transformer, make_column_selector
from sklearn.metrics import accuracy_score

from sklearn.naive_bayes import BernoulliNB
```

```python
df = pd.read_csv(r"C:\Users\PGCP-AI.STUDENTSDC\Desktop\Machine_Learning\Cases\Telecom\Telecom.csv")
df.head()
```

```python
X ,y = df.drop('Response',axis=1), df['Response']
X_train,X_test,y_train,y_test = train_test_split(X,y,test_size=0.3,random_state=26,stratify=y)
```

```python
ohe = OneHotEncoder(sparse_output = False, drop='first').set_output(transform='pandas')
nb = BernoulliNB()
pipe = Pipeline([('OHE',ohe),('NB',nb)])
pipe.fit(X_train,y_train)
```

```python
y_pred = pipe.predict(X_test)
accuracy_score(y_test,y_pred)
```

```python
pipe.predict(X_test)
```

```python
pipe.predict_proba(X_test)[:5]
```

```python
pd.DataFrame((pipe.predict_proba(X_test)).round(2)*100,columns=['No Probability','Yes Probability']).head(5)
```

# Exercise

```python
df = pd.read_csv(r"C:\Users\PGCP-AI.STUDENTSDC\Desktop\Machine_Learning\Cases\Cancer\Cancer.csv",index_col=0)
df.head()
```

```python
df.shape
```

```python
df.info()
```

```python
df.dropna(inplace=True)
df.isnull().sum()
```

```python
df.drop_duplicates(inplace=True)
df.duplicated().sum()
```

```python
X ,y = df.drop('Class',axis=1) , df['Class']
X_train,X_test,y_train,y_test = train_test_split(X,y,test_size=0.3,random_state=26,stratify=y)
```

```python
from sklearn.impute import SimpleImputer

imputer = SimpleImputer(strategy='most_frequent').set_output(transform='pandas')

ohe = OneHotEncoder(sparse_output=False,drop='first').set_output(transform='pandas')

ct = make_column_transformer((ohe,make_column_selector(dtype_include='object'))).set_output(transform='pandas')
nb = BernoulliNB()

pipe = Pipeline([('IMP',imputer),('OHE',ct),('NB',nb)])
pipe.fit(X_train,y_train)
```

```python
y_pred = pipe.predict(X_test)
accuracy_score(y_test,y_pred)
```

# GaussianNB

```python
df = pd.read_csv(r"C:\Users\PGCP-AI.STUDENTSDC\Desktop\Machine_Learning\Cases\Vehicle Silhouettes\Vehicle.csv")
df.head()
```

```python
df.isna().sum().sum()
```

```python
X ,y = df.drop('Class',axis=1) , df['Class']
X_train,X_test,y_train,y_test = train_test_split(X,y,test_size=0.3,random_state=26,stratify=y)
```

```python
from sklearn.preprocessing import StandardScaler
from sklearn.naive_bayes import GaussianNB
scaler = StandardScaler()
gb = GaussianNB()

pipe = Pipeline([('SCL',scaler),('GB',gb)])
pipe.fit(X_train,y_train)
```

```python
y_pred = pipe.predict(X_test)
accuracy_score(y_test,y_pred)
```
