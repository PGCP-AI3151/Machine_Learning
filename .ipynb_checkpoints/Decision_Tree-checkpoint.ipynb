# ----- Classification -----

```python
import pandas as pd
import numpy as np
from sklearn.model_selection import train_test_split
from sklearn.tree import DecisionTreeClassifier,plot_tree
from sklearn.metrics import accuracy_score
from tqdm import tqdm
```

```python
df = pd.read_csv(r"C:\Users\PGCP-AI.STUDENTSDC\Desktop\Machine_Learning\Cases\Wisconsin\BreastCancer.csv",index_col=0)
df.sample(5)
```

```python
X , y = df.drop('Class',axis=1) , df['Class']
X_train,X_test,y_train,y_test = train_test_split(X,y,test_size=0.3,random_state=26,stratify=y)
```

```python
model = DecisionTreeClassifier(random_state=26, max_depth=3)
model.fit(X_train,y_train)
```

```python
import matplotlib.pyplot as plt
plt.figure(figsize=(15,10))
plot_tree(model,feature_names=list(X.columns),
          class_names=['Begin','Malignant'], filled=True, fontsize=10)
plt.show()
```

```python
depths = [None,2,3,4,5,6,7]
min_split = [2,5,10,20,50]   # int -> exact value
min_leaf = [1,5,10,20,50]
score = []                       
for d in depths:
    for s in min_split:
        for l in min_leaf:
            model = DecisionTreeClassifier(random_state=26, max_depth=d,min_samples_split=s,min_samples_leaf=l)
            model.fit(X_train,y_train)
            y_pred = model.predict(X_test)
            score.append([d,s,l,accuracy_score(y_test,y_pred)])
df_scores = pd.DataFrame(score, columns = ['Depth','Min Sample Split','Min Sample Leaf','Accuracy'])
df_scores.sort_values('Accuracy',ascending = False)
```

```python
depths = [None,2,3,4,5,6,7]
min_split = [2,0.05,0.01,0.1,0.2,0.3]   # float -> fraction % of training data as min_sample split
min_leaf = [1,0.05,0.01,0.01,0.2,0.3]
score = []                       
for d in depths:
    for s in min_split:
        for l in min_leaf:
            model = DecisionTreeClassifier(random_state=26, max_depth=d,min_samples_split=s,min_samples_leaf=l)
            model.fit(X_train,y_train)
            y_pred = model.predict(X_test)
            score.append([d,s,l,accuracy_score(y_test,y_pred)])
df_scores = pd.DataFrame(score, columns = ['Depth','Min Sample Split','Min Sample Leaf','Accuracy'])
df_scores.sort_values('Accuracy',ascending = False)
```

# Exercise

```python
df = pd.read_csv(r"C:\Users\PGCP-AI.STUDENTSDC\Desktop\Machine_Learning\Cases\human-resources-analytics\HR_comma_sep.csv")
df.sample(5)
```

```python
X , y = df.drop('left',axis=1), df['left']
X_train,X_test,y_train,y_test = train_test_split(X,y,test_size=0.3,random_state=26,stratify=y)
```

```python
from sklearn.preprocessing import OneHotEncoder, StandardScaler
from sklearn.pipeline import Pipeline
from sklearn.compose import make_column_transformer, make_column_selector


encoder = OneHotEncoder(sparse_output=False, drop='first').set_output(transform='pandas')
ct = make_column_transformer((encoder,make_column_selector(dtype_include='object')),
                             ('passthrough',make_column_selector(dtype_include='number')),
                             verbose_feature_names_out = False).set_output(transform='pandas')

scaler = StandardScaler()
```

```python
depths = [None,2,3,4,5,6,7]
min_split = [2,0.05,0.01,0.1,0.2,0.3]   # float -> fraction % of training data as min_sample split
min_leaf = [1,0.05,0.01,0.01,0.2,0.3]
score = []                       
for d in tqdm(depths):
    for s in min_split:
        for l in min_leaf:
            model = DecisionTreeClassifier(random_state=26, max_depth=d,min_samples_split=s,min_samples_leaf=l)
            pipe = Pipeline([('CT',ct),('TREE',model)])
            pipe.fit(X_train,y_train)
            y_pred = pipe.predict(X_test)
            score.append([d,s,l,accuracy_score(y_test,y_pred)])
df_scores = pd.DataFrame(score, columns = ['Depth','Min Sample Split','Min Sample Leaf','Accuracy'])
df_scores.sort_values('Accuracy',ascending = False,ignore_index=True)
```

```python
best_tree = DecisionTreeClassifier(random_state=26,max_depth=7,min_samples_split=2,min_samples_leaf=1)
X_ohe = ct.fit_transform(X)
best_tree.fit(X_ohe,y)

df_imp = pd.DataFrame({'Features':list(X_ohe.columns),
                       'Importance':best_tree.feature_importances_})
df_imp.sort_values('Importance',inplace=True)
plt.barh(df_imp['Features'], df_imp['Importance'])
plt.title('Feature Importance Plot')
plt.xlabel('Features')
plt.ylabel('Importance')
plt.show()
```

# ----- Regression -----

```python
from sklearn.tree import DecisionTreeRegressor
```

```python
df = pd.read_csv(r"C:\Users\PGCP-AI.STUDENTSDC\Desktop\Machine_Learning\Datasets\Housing.csv")
df.sample(5)
```

```python
X , y = df.drop('price',axis=1), df['price']
X_train,X_test,y_train,y_test = train_test_split(X,y,test_size=0.3,random_state=26)
```

```python
tree = DecisionTreeRegressor(random_state=26,max_depth=2)
pipe = Pipeline([('OHE',ct),('TREE',tree)])
pipe.fit(X_train,y_train)
```

```python
from sklearn.metrics import r2_score
y_pred = pipe.predict(X_test)
r2_score(y_test,y_pred)
```

```python
plt.figure(figsize=(15,10))
plot_tree(tree,feature_names=list(X.columns),
           filled=True, fontsize=10)
plt.show()
```

```python
depths = [None,2,3,4,5,6,7]
min_split = [2,0.05,0.01,0.1,0.2,0.3]   
min_leaf = [1,0.05,0.01,0.01,0.2,0.3]
score = []                       
for d in tqdm(depths):
    for s in min_split:
        for l in min_leaf:
            tree = DecisionTreeRegressor(random_state=26, max_depth=d,min_samples_split=s,min_samples_leaf=l)
            pipe = Pipeline([('OHE',ct),('TREE',tree)])
            pipe.fit(X_train,y_train)
            y_pred = pipe.predict(X_test)
            score.append([d,s,l,r2_score(y_test,y_pred)])
df_scores = pd.DataFrame(score, columns = ['Depth','Min Sample Split','Min Sample Leaf','R2 Score'])
df_scores.sort_values('R2 Score',ascending = False,ignore_index=True)
```

```python
best_tree = DecisionTreeRegressor(random_state=26,max_depth=None,min_samples_split=2,min_samples_leaf=0.05)
X_ohe = ct.fit_transform(X)
best_tree.fit(X_ohe,y)                                                                      
df_imp = pd.DataFrame({'Features':X_ohe.columns,
                       'Importance':best_tree.feature_importances_})
df_imp.sort_values('Importance',inplace=True)
plt.barh(df_imp['Features'], df_imp['Importance'])
plt.title('Feature Importance Plot')
plt.xlabel('Features')
plt.ylabel('Importance')
plt.show()
```
