```python
import pandas as pd
```

```python
job = pd.read_csv(r"C:\Users\PGCP-AI.STUDENTSDC\Desktop\Machine_Learning\Datasets\JobSalary2.csv",index_col = 0)
job
```

## Mean Imputator

```python
mu = dict(job.mean())
job.fillna(mu)
```

## Median Imputator

```python
mu = dict(job.median())
job.fillna(mu)
```

# Simple Imputer

```python
from sklearn.impute import SimpleImputer
```

```python
imputer = SimpleImputer(strategy='mean').set_output(transform='pandas') # default strategy='mean'
imputer.fit(job) # calculate statistics only
print(imputer.statistics_)
imputer.transform(job) # fills the Nan with statistic values
```

```python
imputer.fit_transform(job)  # do both fit and transform simultaneously 
```

```python
imputer = SimpleImputer(strategy = 'median').set_output(transform='pandas')
imputer.fit_transform(job)
print(imputer.statistics_)
```

```python
imputer.fit_transform(job)
```
