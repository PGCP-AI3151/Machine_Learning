```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
from sklearn.linear_model import LinearRegression
from sklearn.preprocessing import PolynomialFeatures
from sklearn.model_selection import train_test_split
from sklearn.metrics import root_mean_squared_error
```

```python
df = pd.read_csv(r"C:\Users\PGCP-AI.STUDENTSDC\Desktop\Machine_Learning\Datasets\pizza.csv")

X = df[['Promote']]
y = df['Sales']

poly = PolynomialFeatures(degree=3,include_bias=False).set_output(transform='pandas')
X_poly = poly.fit_transform(X)
X_poly
```

```python
X_train,X_test,y_train,y_test = train_test_split(X_poly,y,random_state=26,train_size=0.7)
model = LinearRegression()
model.fit(X_train,y_train)
y_pred = model.predict(X_test)
root_mean_squared_error(y_test,y_pred)
```

```python
print(model.coef_)
print(model.intercept_)
```
